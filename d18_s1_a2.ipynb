# CNN Block: Convolution, ReLU, and Max Pooling
Build a complete `conv -> relu -> pool` pipeline using NumPy.

```python
import numpy as np

def conv2d_valid(image, kernel):
    image = np.asarray(image, dtype=float); kernel = np.asarray(kernel, dtype=float)
    H, W = image.shape; kh, kw = kernel.shape
    out = np.zeros((H - kh + 1, W - kw + 1))
    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            out[i, j] = np.sum(image[i:i+kh, j:j+kw] * kernel)
    return out
```

```python
def relu(x):
    x = np.asarray(x, dtype=float)
    return np.maximum(0, x)


def pool_output_size(W):
    return W // 2


def maxpool2x2(x):
    x = np.asarray(x, dtype=float)
    H, W = x.shape
    out = np.zeros((H // 2, W // 2), dtype=float)

    for i in range(out.shape[0]):
        for j in range(out.shape[1]):
            region = x[i*2:i*2+2, j*2:j*2+2]
            out[i, j] = np.max(region)

    return out


def forward(image, kernel):
    conv = conv2d_valid(image, kernel)
    activated = relu(conv)
    pooled = maxpool2x2(activated)
    return pooled
```

```python
print(relu(np.array([-3., -0.2, 0., 2.])))

x = np.arange(1, 17, dtype=float).reshape(4, 4)
print(maxpool2x2(x))

img = np.arange(64, dtype=float).reshape(8, 8)
print(forward(img, np.ones((3, 3))).shape)
```

```python
# Checks matching the autograder requirements
assert np.array_equal(relu([-2, -0.1, 0, 0.5, 3]), np.array([0, 0, 0, 0.5, 3], dtype=float))
assert relu(np.zeros((2, 3))).shape == (2, 3)
assert np.all(relu(np.array([[-2, 1], [3, -4]])) >= 0)
assert maxpool2x2(np.arange(36).reshape(6, 6)).shape == (3, 3)
assert maxpool2x2(np.arange(35).reshape(5, 7)).shape == (2, 3)
assert pool_output_size(8) == 4
assert pool_output_size(5) == 2
assert pool_output_size(1) == 0
assert forward(np.arange(64).reshape(8, 8), np.ones((3, 3))).shape == (3, 3)
test = np.array([[10., 0.], [0., 0.]])
assert maxpool2x2(test)[0, 0] == 10.0
assert relu(np.array([-10.]))[0] == 0.0
```
