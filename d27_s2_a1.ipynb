# DDPM forward process

```python
import numpy as np

def linear_beta_schedule(T, beta_start=1e-4, beta_end=0.02):
    return np.linspace(beta_start, beta_end, T)

def compute_alpha_bar(betas):
    return np.cumprod(1.0 - np.asarray(betas, dtype=float))

def q_sample(x0, t, alpha_bars, noise):
    ab = alpha_bars[t]
    return np.sqrt(ab) * x0 + np.sqrt(1.0 - ab) * noise

def signal_noise_rates(alpha_bars, t):
    ab = alpha_bars[t]
    return (float(np.sqrt(ab)), float(np.sqrt(1.0 - ab)))
```

```python
betas = linear_beta_schedule(1000, 1e-4, 0.02)
ab = compute_alpha_bar(betas)
print(ab.shape, round(float(ab[0]), 4))     # (1000,) 0.9999
print(round(float(ab[-1]), 6))              # ~3.9e-05
print(q_sample(np.array([1.,2.,3.]), 0, np.array([0.64]), np.array([1.,1.,1.])))  # [1.4 2.2 3.0]
print(signal_noise_rates(np.array([0.36]), 0))                                    # (0.6, 0.8)
print(q_sample(np.array([2.,-1.]), 30, ab, np.zeros(2)))
rng = np.random.default_rng(0)
xT = q_sample(rng.normal(size=(100000,)), 999, ab, rng.standard_normal(100000))
print(round(xT.mean(), 3), round(xT.var(), 3))  # ~0, ~1
```
