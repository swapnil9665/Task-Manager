```python
import numpy as np

z = np.array([-2.0, -0.5, 0.0, 0.5, 2.0])

def sigmoid(z):
    z = np.asarray(z, dtype=float)
    out = np.empty_like(z)
    pos = z >= 0
    out[pos] = 1.0 / (1.0 + np.exp(-z[pos]))
    exp_z = np.exp(z[~pos])
    out[~pos] = exp_z / (1.0 + exp_z)
    return out

def relu(z):
    return np.maximum(0, np.asarray(z))

def tanh_act(z):
    return np.tanh(z)

def sigmoid_grad(z):
    s = sigmoid(z)
    return s * (1.0 - s)

def relu_grad(z):
    return (np.asarray(z) > 0).astype(float)

def tanh_grad(z):
    t = np.tanh(z)
    return 1.0 - t**2

def bce_loss(y_true, y_pred, eps=1e-12):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    p = np.clip(y_pred, eps, 1.0 - eps)
    loss = -np.mean(y_true * np.log(p) + (1.0 - y_true) * np.log(1.0 - p))
    return float(loss)
```

```python
z = np.array([-2.0, 0.0, 2.0])
print(np.round(sigmoid(z), 4))
print(np.round(sigmoid_grad(z), 4))
print(relu(np.array([-1.0, 2.0])))
print(np.round(tanh_grad(0.0), 4))
print(round(bce_loss([1,0,1,0], [0.9,0.1,0.8,0.2]), 4))
print(round(bce_loss([1,0], [0.0,1.0]), 4))

# Stability checks
print(sigmoid(np.array([-1000.0, 1000.0])))
print(np.isfinite(sigmoid(np.array([-1000.0, 1000.0]))).all())
```
