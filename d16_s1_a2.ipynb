```python

import numpy as np

from sklearn.datasets import load_digits
from sklearn.decomposition import PCA


# ==========================================
# STEP 1: LOAD DATASET
# ==========================================

X = load_digits().data

print("Dataset Shape:", X.shape)


# ==========================================
# STEP 2: PCA EXPLAINED VARIANCE RATIOS
# ==========================================

def pca_ratios(X):
    pca = PCA()
    pca.fit(X)

    return pca.explained_variance_ratio_


# ==========================================
# STEP 3: CUMULATIVE EXPLAINED VARIANCE
# ==========================================

def cumulative(ratios):
    return np.cumsum(ratios)


# ==========================================
# STEP 4: CHOOSE NUMBER OF COMPONENTS
# ==========================================

def choose_k(ratios, threshold=0.9):

    cum = cumulative(ratios)

    indices = np.where(cum >= threshold)[0]

    if len(indices) == 0:
        return len(ratios)

    return int(indices[0] + 1)


# ==========================================
# STEP 5: PROJECT DATA ONTO TOP K COMPONENTS
# ==========================================

def project(X, k):

    pca = PCA(n_components=k)

    return pca.fit_transform(X)


# ==========================================
# STEP 6: VERIFY PCA RATIOS
# ==========================================

r = pca_ratios(X)

print("\nFirst 3 Explained Variance Ratios:")
print(r[:3].round(3))

print("\nSum of Explained Variance Ratios:")
print(r.sum())


# ==========================================
# STEP 7: VERIFY CUMULATIVE VARIANCE
# ==========================================

c = cumulative(r)

print("\nFirst 5 Cumulative Ratios:")
print(c[:5].round(3))

print("\nFinal Cumulative Variance:")
print(c[-1])


# ==========================================
# STEP 8: FIND K FOR 90% VARIANCE
# ==========================================

k = choose_k(r, 0.9)

print("\nComponents Required for 90% Variance:")
print(k)


# ==========================================
# STEP 9: PROJECT DATA
# ==========================================

X_projected = project(X, k)

print("\nProjected Data Shape:")
print(X_projected.shape)


# ==========================================
# STEP 10: TEST KNOWN INPUTS
# ==========================================

test_ratios = np.array([0.5, 0.3, 0.15, 0.05])

print("\nTest 1:")
print(choose_k(test_ratios, 0.9))  # 3


# Exact boundary test
test_ratios_2 = np.array([0.5, 0.25, 0.25])

print("\nTest 2:")
print(choose_k(test_ratios_2, 0.75))  # 2


# Full variance test
print("\nTest 3 - Threshold 1.0:")
print(choose_k(r, 1.0))  # 64


# ==========================================
# STEP 11: VERIFY DIMENSIONALITY REDUCTION
# ==========================================

print("\nOriginal Features:", X.shape[1])
print("Reduced Features:", k)

print("Variance Retained:", c[k - 1])

assert k < X.shape[1]
assert c[k - 1] >= 0.9
```
