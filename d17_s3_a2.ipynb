```python
import numpy as np
from sklearn.datasets import load_digits

_, y = load_digits(return_X_y=True)
y_true = np.array([0, 0, 1, 1, 2, 2])
y_pred = np.array([0, 1, 1, 1, 2, 0])

def one_hot(y, num_classes):
    y = np.asarray(y, dtype=int)
    encoded = np.zeros((len(y), num_classes), dtype=int)
    encoded[np.arange(len(y)), y] = 1
    return encoded

def confusion_matrix(y_true, y_pred, num_classes):
    y_true = np.asarray(y_true, dtype=int)
    y_pred = np.asarray(y_pred, dtype=int)
    cm = np.zeros((num_classes, num_classes), dtype=int)
    np.add.at(cm, (y_true, y_pred), 1)
    return cm

def overall_accuracy(cm):
    cm = np.asarray(cm)
    total = cm.sum()
    if total == 0:
        return 0.0
    return float(np.trace(cm) / total)

def per_class_accuracy(cm):
    cm = np.asarray(cm)
    row_sums = cm.sum(axis=1)
    recalls = np.zeros(cm.shape[0], dtype=float)
    np.divide(np.diag(cm), row_sums, out=recalls, where=row_sums != 0)
    return recalls

def most_confused_pair(cm):
    cm = np.asarray(cm)
    off_diagonal = cm.copy()
    np.fill_diagonal(off_diagonal, -1)
    index = np.argmax(off_diagonal)
    true_class, predicted_class = np.unravel_index(index, cm.shape)
    return int(true_class), int(predicted_class)
```

```python
print(one_hot([0, 2, 1], 3))
cm = confusion_matrix(y_true, y_pred, 3)
print(cm)
print(round(overall_accuracy(cm), 3))
print(np.round(per_class_accuracy(cm), 3))
print(most_confused_pair(cm))
```
