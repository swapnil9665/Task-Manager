# RNN Cell and Forward Pass Assignment
Notebook: `d21_s1_a1.ipynb`


```python
import numpy as np

D, H = 2, 3
Wxh = np.array([[0.10, 0.20], [0.00, -0.10], [0.30, 0.10]])   # (H, D)
Whh = np.array([[0.05, 0.00, 0.10], [0.10, 0.05, 0.00], [0.00, 0.10, 0.05]])  # (H, H)
b   = np.array([0.00, 0.10, -0.10])                            # (H,)
h0  = np.zeros(3)                                              # (H,)
X   = np.array([[1.0, 0.5], [-0.5, 1.0]])                      # (T, D)
```

```python
def rnn_cell_step(x, h_prev, Wxh, Whh, b):
    return np.tanh(Wxh @ x + Whh @ h_prev + b)


def rnn_forward(X, h0, Wxh, Whh, b):
    h_prev = h0.copy()
    hidden_states = []

    for x in X:
        h = rnn_cell_step(x, h_prev, Wxh, Whh, b)
        hidden_states.append(h)
        h_prev = h

    return np.array(hidden_states)


def count_params(D, H):
    return H * (D + H + 1)
```

```python
h1 = rnn_cell_step(X[0], h0, Wxh, Whh, b)
print(np.round(h1, 4))

hs = rnn_forward(X, h0, Wxh, Whh, b)
print(hs.shape)
print(np.round(hs[-1], 4))

print(count_params(2, 3))
print(count_params(10, 20))
```
