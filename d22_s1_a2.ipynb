# Greedy Decoding, Fixed Context, and Sequence Accuracy


```python
import numpy as np

def onehot_logits(tokens, V=6):
    L = np.full((len(tokens), V), -1.0)
    for t, tok in enumerate(tokens):
        L[t, tok] = 5.0
    return L

def greedy_decode(logits, eos_id=2, max_len=None):
    T = len(logits)
    limit = T if max_len is None else min(T, max_len)
    result = []
    for t in range(limit):
        token = int(np.argmax(logits[t]))
        if token == eos_id:
            break
        result.append(token)
    return result

def decode_batch(logits_list, eos_id=2, max_len=None):
    return [greedy_decode(logits, eos_id, max_len) for logits in logits_list]

def summarize_context(X):
    return np.mean(X, axis=0)

def exact_match(pred, target):
    return 1.0 if pred == target else 0.0

def sequence_accuracy(preds, targets):
    if len(preds) == 0:
        return 0.0
    return float(np.mean([exact_match(p, t) for p, t in zip(preds, targets)]))

logits = onehot_logits([3, 4, 5, 2, 4])
print(greedy_decode(onehot_logits([3, 4, 5, 2, 4])))
print(greedy_decode(onehot_logits([3, 4, 5, 3])))
X = np.random.default_rng(0).normal(size=(10, 8))
print(summarize_context(X).shape)
print(sequence_accuracy([[3, 4], [5, 6]], [[3, 4], [5, 9]]))
```
