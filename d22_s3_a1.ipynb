# Time-Series Windowing and Persistence Baseline


```python
import numpy as np

rng = np.random.RandomState(22)
t = np.arange(240)
series = 50 + 0.1 * t + 8 * np.sin(2 * np.pi * t / 24) + rng.randn(240)

def make_windows(series, lookback, horizon=1):
    series = np.asarray(series)
    N = len(series) - lookback - horizon + 1
    X = np.array([series[i:i + lookback] for i in range(N)])
    y = np.array([series[i + lookback:i + lookback + horizon] for i in range(N)])
    return X, y

def chrono_split(X, y, test_frac=0.2):
    n_test = max(1, round(len(X) * test_frac))
    split = len(X) - n_test
    return X[:split], X[split:], y[:split], y[split:]

def naive_forecast(X, horizon=1):
    X = np.asarray(X)
    return np.repeat(X[:, -1][:, None], horizon, axis=1)

def rmse(y_true, y_pred):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    return float(np.sqrt(np.mean((y_true - y_pred) ** 2)))

def mae(y_true, y_pred):
    y_true = np.asarray(y_true).ravel()
    y_pred = np.asarray(y_pred).ravel()
    return float(np.mean(np.abs(y_true - y_pred)))

s = np.arange(20.)
X, y = make_windows(s, 4, 1)
print(X.shape, y.shape)
print(X[0], y[0].ravel())
Xtr, Xte, ytr, yte = chrono_split(X, y, 0.25)
print(len(Xtr), len(Xte))
print(naive_forecast(np.array([[1., 2., 3.]]), 2))
print(round(rmse([0, 0, 0], [3, 4, 0]), 4), round(mae([0, 0, 0], [3, 4, 0]), 4))
```
