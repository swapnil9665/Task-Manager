# BPTT Gradient Vanishing and Exploding
Notebook: `d21_s2_a1.ipynb`


```python
import numpy as np

W = np.array([[0.5]])
g0 = np.array([1.0])
```

```python
def spectral_radius(W):
    W = np.asarray(W, dtype=float)
    eigenvalues = np.linalg.eigvals(W)
    return float(np.max(np.abs(eigenvalues)))


def propagate_grad_norms(W, g0, k_max):
    W = np.asarray(W, dtype=float)
    g = np.asarray(g0, dtype=float).copy()
    norms = [float(np.linalg.norm(g))]

    for _ in range(k_max):
        g = W @ g
        norms.append(float(np.linalg.norm(g)))

    return norms


def log_norm_slope(norms):
    norms = np.asarray(norms, dtype=float)
    k = np.arange(len(norms))
    return float(np.polyfit(k, np.log(norms), 1)[0])
```

```python
print(round(spectral_radius(np.diag([0.5, 0.3])), 4))
print(round(spectral_radius([[0, 1], [1, 0]]), 4))
norms = propagate_grad_norms([[0.5]], [1.0], 5)
print([round(x, 4) for x in norms])
print(round(log_norm_slope(norms), 4))
print(round(log_norm_slope(propagate_grad_norms([[1.2]], [1.0], 20)), 4))
```

```python
# Additional autograder-style checks
assert np.isclose(spectral_radius(np.diag([0.4, -0.7])), 0.7)
test_norms = propagate_grad_norms(np.eye(2), [3.0, 4.0], 4)
assert test_norms[0] == 5.0 and len(test_norms) == 5
assert log_norm_slope(propagate_grad_norms([[0.7]], [1.0], 20)) < 0
assert log_norm_slope(propagate_grad_norms([[1.2]], [1.0], 20)) > 0
W_test = np.diag([0.8, 0.3])
dominant_norms = propagate_grad_norms(W_test, [1.0, 0.0], 20)
assert np.isclose(log_norm_slope(dominant_norms), np.log(spectral_radius(W_test)))
print('All additional checks passed.')
```
