# CNN Convolution from Scratch
Manual valid 2D cross-correlation and a Prewitt vertical-edge detector.

```python
import numpy as np

img = np.arange(25, dtype=float).reshape(5, 5)
edge = np.zeros((5, 6))
edge[:, 3:] = 10.0
```

```python
def conv_output_size(W, K, P=0, S=1):
    return (W - K + 2 * P) // S + 1


def conv2d_valid(image, kernel):
    H, W = image.shape
    kh, kw = kernel.shape
    output = np.zeros((H - kh + 1, W - kw + 1), dtype=float)

    for i in range(H - kh + 1):
        for j in range(W - kw + 1):
            region = image[i:i + kh, j:j + kw]
            output[i, j] = np.sum(region * kernel)

    return output


def vertical_edge_kernel():
    return np.array([[-1, 0, 1],
                     [-1, 0, 1],
                     [-1, 0, 1]], dtype=float)


def detect_vertical_edges(image):
    return conv2d_valid(image, vertical_edge_kernel())
```

```python
print(conv_output_size(5, 3))
print(conv_output_size(7, 3, 1, 2))
print(conv2d_valid(np.ones((4, 4)), np.ones((3, 3))))
print(detect_vertical_edges(edge))
```

```python
# Extra checks matching the grading requirements
assert conv_output_size(5, 3) == 3
assert conv_output_size(28, 5) == 24
assert conv_output_size(7, 3, 1, 2) == 4
assert conv_output_size(32, 5, 2, 1) == 32
assert conv2d_valid(np.zeros((5, 7)), np.ones((3, 3))).shape == (3, 5)
assert np.isclose(vertical_edge_kernel().sum(), 0.0)
assert np.allclose(detect_vertical_edges(np.ones((5, 5))), 0.0)
```
