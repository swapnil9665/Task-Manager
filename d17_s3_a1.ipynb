```python
import numpy as np
from sklearn.datasets import load_digits
from sklearn.neural_network import MLPClassifier

X, y = load_digits(return_X_y=True)
X = X / 16.0
clf = MLPClassifier(hidden_layer_sizes=(30,), activation="relu", max_iter=300, random_state=1).fit(X, y)
params = list(zip(clf.coefs_, clf.intercepts_))

def relu(Z):
    return np.maximum(0, Z)

def softmax(Z):
    Z = np.asarray(Z, dtype=float)
    shifted = Z - np.max(Z, axis=1, keepdims=True)
    exp_Z = np.exp(shifted)
    return exp_Z / np.sum(exp_Z, axis=1, keepdims=True)

def mlp_forward(X, params):
    A = np.asarray(X, dtype=float)
    for i, (W, b) in enumerate(params):
        Z = A @ W + b
        A = relu(Z) if i < len(params) - 1 else softmax(Z)
    return A

def predict(X, params):
    return np.argmax(mlp_forward(X, params), axis=1)

def accuracy(y_true, y_pred):
    return float(np.mean(np.asarray(y_true) == np.asarray(y_pred)))
```

```python
probs = mlp_forward(X, params)
print(probs.shape)
print(np.allclose(probs.sum(axis=1), 1.0))
preds = predict(X, params)
print(preds[:8])
print(round(accuracy(y, preds), 3))
print(np.array_equal(preds, clf.predict(X)))
print(np.allclose(probs, clf.predict_proba(X)))
```
