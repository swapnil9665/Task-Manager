# Teacher Forcing Decoder Inputs, Targets, and Loss Mask


```python
import numpy as np

targets = [[7, 8, 9], [4, 5], [3, 6, 6, 7]]
BOS, EOS, PAD = 1, 2, 0

def make_decoder_io(target, bos_id=1, eos_id=2):
    return {
        "input": [bos_id] + target,
        "target": target + [eos_id]
    }

def pad(seq, length, pad_id=0):
    seq = seq[:length]
    return seq + [pad_id] * (length - len(seq))

def loss_mask(dec_target, pad_id=0):
    dec_target = np.array(dec_target)
    return (dec_target != pad_id).astype(float)

def build_batch(targets, bos_id=1, eos_id=2, pad_id=0):
    decoder_pairs = [make_decoder_io(t, bos_id, eos_id) for t in targets]
    L = max(len(t) + 1 for t in targets)

    dec_in = np.array([pad(x["input"], L, pad_id) for x in decoder_pairs])
    dec_out = np.array([pad(x["target"], L, pad_id) for x in decoder_pairs])
    mask = loss_mask(dec_out, pad_id)

    return dec_in, dec_out, mask

print(make_decoder_io([7, 8, 9]))
di, do, m = build_batch(targets)
print(di.shape, do.shape)
print(do[1].tolist())
print(m[1].tolist())
print(m.sum(axis=1).tolist())
```
