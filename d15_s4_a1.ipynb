```python
import numpy as np
from sklearn.datasets import make_regression

def make_data(seed=0):
    return make_regression(
        n_samples=200,
        n_features=5,
        n_informative=5,
        noise=10.0,
        random_state=seed
    )

def add_bias(X):
    X = np.asarray(X)
    ones = np.ones((X.shape[0], 1), dtype=X.dtype)
    return np.hstack([ones, X])

def fit_ols(X, y):
    Xb = add_bias(X)
    y = np.asarray(y)
    theta = np.linalg.inv(Xb.T @ Xb) @ (Xb.T @ y)
    return theta

def predict(X, theta):
    return add_bias(X) @ theta

def mse_scratch(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    return float(np.mean((y_true - y_pred) ** 2))

def r2_scratch(y_true, y_pred):
    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)
    ss_res = np.sum((y_true - y_pred) ** 2)
    ss_tot = np.sum((y_true - np.mean(y_true)) ** 2)
    return float(1.0 - ss_res / ss_tot)
```

```python
import numpy as np

rng = np.random.default_rng(7)
Xk = rng.standard_normal((150, 4))
true = np.array([5.0, 2.0, -3.0, 0.5, 1.5])
yk = Xk @ true[1:] + true[0]
theta = fit_ols(Xk, yk)
print(np.round(theta, 3))
print(round(mse_scratch(yk, predict(Xk, theta)), 8))
print(round(r2_scratch(yk, predict(Xk, theta)), 8))
print(round(mse_scratch([0.0, 0.0], [3.0, 4.0]), 3))
```
