```python
import time
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.naive_bayes import GaussianNB

def load_data():
    d = load_wine()
    return d.data, d.target

def model_zoo():
    return {
        "logreg": LogisticRegression(max_iter=5000),
        "tree": DecisionTreeClassifier(random_state=0),
        "forest": RandomForestClassifier(n_estimators=100, random_state=0),
        "knn": KNeighborsClassifier(n_neighbors=5),
        "nb": GaussianNB()
    }

def benchmark(models, X, y, seed=0):
    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.25, random_state=seed, stratify=y
    )
    results = []
    for name, model in models.items():
        t0 = time.perf_counter()
        model.fit(X_train, y_train)
        fit_time = time.perf_counter() - t0
        
        preds = model.predict(X_test)
        acc = round(float(accuracy_score(y_test, preds)), 4)
        results.append({
            "name": name,
            "accuracy": acc,
            "fit_time": float(fit_time)
        })
    return results

def leaderboard(results):
    return sorted(results, key=lambda r: (-r["accuracy"], r["fit_time"]))
```

```python
X, y = load_data()
rows = benchmark(model_zoo(), X, y, 0)
for r in leaderboard(rows):
    print(r["name"], r["accuracy"], round(r["fit_time"], 4))
```
