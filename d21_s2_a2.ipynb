# Global-Norm Gradient Clipping
Notebook: `d21_s2_a2.ipynb`


```python
import numpy as np

grads = [np.array([3.0, 4.0]), np.array([12.0, 0.0])]
max_norm = 6.5
```

```python
def global_norm(grads):
    return float(np.sqrt(sum(np.sum(np.asarray(g, dtype=float) ** 2) for g in grads)))


def did_clip(grads, max_norm):
    return bool(global_norm(grads) > max_norm)


def clip_by_global_norm(grads, max_norm):
    norm = global_norm(grads)

    if norm == 0 or norm <= max_norm:
        return [np.asarray(g).copy() for g in grads]

    factor = max_norm / norm
    return [np.asarray(g) * factor for g in grads]
```

```python
grads = [np.array([3.0, 4.0]), np.array([12.0, 0.0])]
print(round(global_norm(grads), 4))
print(did_clip(grads, 6.5))
out = clip_by_global_norm(grads, 6.5)
print([list(a) for a in out])
print(round(global_norm(out), 4))
print(clip_by_global_norm(grads, 100.0))
```

```python
# Additional autograder-style checks
assert np.isclose(global_norm([np.array([3.0, 4.0])]), 5.0)
assert np.isclose(global_norm(grads), 13.0)
assert did_clip(grads, 6.5) is True
assert did_clip(grads, 20.0) is False
unchanged = clip_by_global_norm(grads, 100.0)
assert all(np.array_equal(a, b) for a, b in zip(unchanged, grads))
clipped = clip_by_global_norm(grads, 6.5)
assert np.isclose(global_norm(clipped), 6.5)
assert all(np.allclose(a, b * 0.5) for a, b in zip(clipped, grads))
zeros = [np.zeros(3), np.zeros((2, 2))]
zero_out = clip_by_global_norm(zeros, 1.0)
assert global_norm(zero_out) == 0.0
print('All additional checks passed.')
```
