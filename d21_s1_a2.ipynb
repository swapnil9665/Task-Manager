# Many-to-One RNN Sequence Classifier
Notebook: `d21_s1_a2.ipynb`


```python
import numpy as np

Wxh = np.array([[0.10, 0.20], [0.00, -0.10], [0.30, 0.10]])
Whh = np.array([[0.05, 0.00, 0.10], [0.10, 0.05, 0.00], [0.00, 0.10, 0.05]])
b = np.array([0.00, 0.10, -0.10])
Why = np.array([[0.20, -0.10, 0.30], [0.00, 0.40, -0.20], [0.10, 0.10, 0.10]])
by = np.array([0.00, 0.10, -0.10])
h0 = np.zeros(3)
X = np.array([[1.0, 0.5], [-0.5, 1.0]])
```

```python
def softmax(z):
    z = np.asarray(z, dtype=float)
    shifted_z = z - np.max(z)
    exp_z = np.exp(shifted_z)
    return exp_z / np.sum(exp_z)


def rnn_final_state(X, h0, Wxh, Whh, b):
    h = h0.copy()
    for x in X:
        h = np.tanh(Wxh @ x + Whh @ h + b)
    return h


def rnn_classify(X, h0, Wxh, Whh, b, Why, by):
    h_T = rnn_final_state(X, h0, Wxh, Whh, b)
    logits = Why @ h_T + by
    return softmax(logits)


def predict(probs):
    return int(np.argmax(probs))


def cross_entropy(probs, y_true):
    return -np.log(probs[y_true])
```

```python
print(np.round(softmax([1.0, 2.0, 3.0]), 4))
probs = rnn_classify(X, h0, Wxh, Whh, b, Why, by)
print(np.round(probs, 4))
print(round(float(probs.sum()), 6))
print(predict(probs))
print(round(cross_entropy(probs, 2), 4))
```

```python
# Additional stability and sanity checks
assert np.allclose(softmax([1000, 1001, 1002]), softmax([1, 2, 3]))
assert np.isclose(softmax([1, 2, 3]).sum(), 1.0)
assert cross_entropy(np.array([0.999999, 0.000001]), 0) < 0.00001
assert rnn_final_state(X, h0, Wxh, Whh, b).shape == (3,)
print('All additional checks passed.')
```
